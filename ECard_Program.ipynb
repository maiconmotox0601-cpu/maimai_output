```python
import random

#一人分のデッキ
cards = ["王", "騎士", "騎士", "騎士", "騎士", "ドラゴン"]

#.copy()でプレイヤーとコンピューターのデッキを作成
player_deck = cards.copy()
comp_deck = cards.copy()

#勝敗判定の関数を定義
def judge(p, c):
    if p == c:
        return "draw"

    if (
        (p == "王" and c == "騎士") or
        (p == "騎士" and c == "ドラゴン") or
        (p == "ドラゴン" and c == "王")
    ):
        return "win"

    return "lose"

#勝ち数
player_score = 0
comp_score = 0

#5ラウンド制のゲームを作る
for round_num in range(1, 6):
    print(f"\n--- {round_num} ラウンド　---") #ラウンド表示

    stop = input("途中でやめますか？(y/n)：")
    if stop.lower() == "y":
        print("ゲームを中断しました。")
        break

    print("あなたのデッキ：", player_deck) #デッキ内容表示

    user = input("出すカードを入力(王/騎士/ドラゴン)：")
    if user not in player_deck: #入力したカードがデッキに存在しないならTrue
        print("そのカードはもうありません。")
        continue

    #使ったカードの削除
    player_deck.remove(user)

    #コンピュータがランダムにカードを選ぶ
    comp = random.choice(comp_deck)

    #コンピュータが使ったカードの削除
    comp_deck.remove(comp)

    print("コンピュータ：", comp)

    #judge関数に渡して勝敗判定、勝ちならスコア+1
    result = judge(user, comp)
    if result == "win":
        print("あなたの勝ち！")
        player_score += 1

    elif result == "lose":
        print("あなたの負け…")
        comp_score += 1

    else:
        print("引き分け")

    #空行
    print()

print("ゲーム終了！")

#最終結果
print("あなたの勝ち数：", player_score)
print("コンピュータの勝ち数：", comp_score)

if player_score > comp_score:
    print("あなたの勝利です！")

elif player_score < comp_score:
    print("あなたの負けです…")

else:
    print("引き分けです")
```
