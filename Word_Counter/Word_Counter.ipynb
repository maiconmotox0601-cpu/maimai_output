```python
#ユーザーが自由に文章を入力できるようにする
#入力した文章をtextに格納
text = input("文章を入力してください：")

#大文字、小文字の違いをなくす
#文字列はイミュータブルだから必ず代入し直す
text = text.lower()

#textの文字列を空白で区切って単語のリストを作る
#リストをwordsに代入
words = text.split()

#前の行(split)で区切ったリスト(words)の長さを返す
print("単語数：", len(words))

#重複を自動で取り除く
unique_words = set(words)
#重複を取り除いた単語数を返す
print("ユニークな単語数：", len(unique_words))

#これから単語を並べ替えて表示するよという見出しのみ
print("単語一覧（ソート済み）：")

#sortedでアルファベット順に並べ替え
for w in sorted(unique_words):
    print("-", w) #ハイフンで箇条書き風に表示

#特定の単語の出現回数を調べる
#小文字でそろえてtargetに代入
target = input("\n出現回数を調べたい単語を入力してください：").lower()
#単語リスト(words)の中にある特定の値(target)の出現回数を数える
print(f"「{target}」の出現回数：", words.count(target))
```
